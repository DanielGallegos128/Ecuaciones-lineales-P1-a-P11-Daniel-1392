Ejercicio 1: $x+8=15$
* Paso 1 (Planteamiento): Tenemos dos números con signos distintos: $-8$ y $+3$.
* Paso 2 (Análisis): Al ser signos diferentes, restamos sus valores absolutos ($8 - 3 = 5$).Paso 3 (Signo): El número con mayor valor absoluto es $8$, que es negativo, por lo tanto el resultado hereda ese signo.

```python
# Problema: x + 8 = 15
valor_derecha = 15
numero_sumando = 8

x = valor_derecha - numero_sumando

print(f"Resultado Ejercicio 1: x = {x}")
```

Ejercicio 3: $3x+4=21$
* Paso 1 (Planteamiento): Eliminamos el paréntesis aplicando la ley de signos ($+ \cdot - = -$).
* Paso 2 (Ecuación): La operación se convierte en $5 - 7$.
* Paso 3 (Solución): Como el $7$ es mayor que el $5$, el resultado es la diferencia negativa entre ellos.

```python
# Problema: 3x + 4 = 21
resultado = 21
termino_independiente = 4
coeficiente_x = 3

x = (resultado - termino_independiente) / coeficiente_x

print(f"Resultado Ejercicio 3: x = {x:.2f}")
```

Ejercicio 5: $6a-2a+a$
* Paso 1 (Planteamiento): Identificamos que ambos números tienen el mismo signo (negativo).
* Paso 2 (Análisis): En signos iguales, los valores se suman ($3 + 4 = 7$).
* Paso 3 (Signo): Se mantiene el signo original de ambos números.

```python
# Problema: 6a - 2a + a
coef1 = 6
coef2 = -2
coef3 = 1

suma_total = coef1 + coef2 + coef3

print(f"Resultado Ejercicio 5: {suma_total}a")
```

Ejercicio 7: $4x+3=2x+11$
* Paso 1 (Planteamiento): Tenemos una deuda de $7$ y un abono de $12$.
* Paso 2 (Análisis): Signos diferentes se restan ($12 - 7 = 5$).
* Paso 3 (Solución): El $12$ es positivo y es el más grande, así que el resultado es positivo.

```python
# Problema: 4x + 3 = 2x + 11
# Agrupamos: (4 - 2)x = 11 - 3
diff_x = 4 - 2
diff_num = 11 - 3

x = diff_num / diff_x

print(f"Resultado Ejercicio 7: x = {x}")
```

Ejercicio 9: $9m+4=5m+20$
* Paso 1 (Planteamiento): Encontramos un signo negativo frente a un paréntesis con otro negativo.
* Paso 2 (Ley de signos): El doble negativo se convierte en positivo ($- \cdot - = +$).
* Paso 3 (Ecuación): Resolvemos la suma resultante: $9 + 3 = 12$.

```python
# Problema: 9m + 4 = 5m + 20
# Agrupamos: (9 - 5)m = 20 - 4
diff_m = 9 - 5
diff_num = 20 - 4

m = diff_num / diff_m

print(f"Resultado Ejercicio 9: m = {m}")
```

Ejercicio 11: $6a+5=2a+21$
* Paso 1 (Planteamiento): Quitamos el paréntesis multiplicando los signos ($+ \cdot - = -$).
* Paso 2 (Ecuación): La operación se transforma en $15 - 20$.
* Paso 3 (Solución): Restamos y mantenemos el signo del número mayor ($20$), dando un resultado negativo.

```python
# Problema: 6a + 5 = 2a + 21
# Agrupamos: (6 - 2)a = 21 - 5
diff_a = 6 - 2
diff_num = 21 - 5

a = diff_num / diff_a

print(f"Resultado Ejercicio 11: a = {a}")
```
